## Excercise - initialize np array

**Specifications**
* create 5 * 5 array
* make all edges values to 1
* make center cell to 9
* make all remaining cell to 0

```python
# Solution 1
res_arr = np.ones((5, 5), dtype="int16")

res_arr[1, 1:-1] = 0
res_arr[-2, 1:-1] = 0
res_arr[1:-1, 1] = 0
res_arr[1:-1, -2] = 0
res_arr[2, 2] = 9

print(res_arr)
```

```python
# Solution 2
res_arr = np.ones((5, 5), dtype="int16")

zero_arr = np.zeros((3, 3), dtype="int16")
zero_arr[1, 1] = 9

res_arr[1:-1, 1:-1] = zero_arr

print(res_arr)
```

## Caution: copy operation

The assignment operator (=) right to left will pass the reference and should be avoided when we need copy of an object. Use copy method to have copy.

```python
# referecing
arr1 = np.array([1, 2, 3, 4, 5, 6, 7])
# both arr1 and aar2 points to same reference by doing the below assignment
arr2 = arr1
arr2[-1] = 10
print(arr1, arr2)
```

```python
# copy object
arr1 = np.array([1, 2, 3, 4, 5, 6, 7])
arr2 = arr1.copy()
arr2[-1] = 10
print(arr1, arr2)
```

## Linear Algebra

```python
# Matrix multiplication
arr1 = np.array([[2, 3, 4],
                 [5, 6, 7]])
arr2 = np.full((3, 4), 1, dtype="int32")
# matrix mulpication ((2*3) * (3*4) => (2*4)) (r1*c1) * (r2*c2) => (r1*c2)
print(np.matmul(arr1, arr2)) 
```

```python
# Identity of a matrix
print(np.linalg.det(np.ones((3, 3))))
print(np.linalg.det(np.identity((3))))
```

## Reorganizing

### Reshape

```python
before = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(f"Before matrix:\n{before}")
print(f"After matrix 1 * 8:\n{before.reshape(1, 8)}")
print(f"After matrix 8 * 1:\n{before.reshape(8, 1)}")
print(f"After matrix 2 * 4:\n{before.reshape(2, 4)}")
print(f"After matrix 4 * 2:\n{before.reshape(4, 2)}")
```

### Vertical stack

```python
before = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
arr1 = np.array([1, 2, 3, 4])
print(np.vstack((arr1, before, arr1, before)))
```

### Horizontal Stack

```python
before = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
arr1 = np.array([[1, 2,], [3, 4]])
print(np.hstack((arr1, before, arr1, before)))
```
